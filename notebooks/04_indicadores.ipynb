# Ejercicio 7: Indicadores y Visualizaciones (Respaldo)
Este notebook actúa como respaldo en código (usando DuckDB y Seaborn) de los indicadores que se configurarán en Metabase. 
Carga los archivos SQL ubicados en `../sql/indicadores/` y ejecuta las consultas directamente sobre la tabla materializada `viajes`.

```python
import duckdb
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Configuración visual
sns.set_theme(style="whitegrid")

# Conectar a la base materializada
DB_PATH = "../data/processed/taxis.duckdb"
con = duckdb.connect(DB_PATH, read_only=True)

def ejecutar_sql(archivo):
    ruta = Path(f"../sql/indicadores/{archivo}")
    query = ruta.read_text()
    return con.execute(query).df()
```

## Indicador 1: Tendencia Mensual de Viajes

```python
df_i1 = ejecutar_sql("01_tendencia_mensual.sql")
plt.figure(figsize=(10, 5))
sns.lineplot(data=df_i1, x="mes", y="total_viajes", hue="tipo", marker="o")
plt.title("Tendencia Mensual de Viajes por Tipo")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Indicador 2: Demanda por Hora del Día

```python
df_i2 = ejecutar_sql("02_demanda_por_hora.sql")
plt.figure(figsize=(12, 5))
sns.barplot(data=df_i2, x="hora_dia", y="total_viajes", hue="tipo", palette="viridis")
plt.title("Demanda de Viajes por Hora del Día")
plt.tight_layout()
plt.show()
```

## Indicador 3: Distribución de Recogidas por Distrito

```python
df_i3 = ejecutar_sql("03_distritos_origen.sql")
plt.figure(figsize=(10, 5))
sns.barplot(data=df_i3, y="distrito", x="total_viajes", hue="tipo")
plt.title("Viajes por Distrito de Origen")
plt.tight_layout()
plt.show()
```

## Indicador 4: Propinas según Método de Pago

```python
df_i4 = ejecutar_sql("04_propinas_pago.sql")
plt.figure(figsize=(8, 5))
sns.barplot(data=df_i4, x="descripcion_pago", y="porcentaje_propina_promedio", palette="magma")
plt.title("% de Propina Promedio por Método de Pago")
plt.tight_layout()
plt.show()
```

## Indicador 5: Tarifa Promedio por Milla

```python
df_i5 = ejecutar_sql("05_tarifa_milla.sql")
plt.figure(figsize=(10, 5))
sns.lineplot(data=df_i5, x="mes", y="tarifa_promedio_por_milla", hue="tipo", marker="s")
plt.title("Tarifa Promedio por Milla Recorrida")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Indicador 6: Impacto de Viajes a Aeropuertos

```python
df_i6 = ejecutar_sql("06_aeropuertos.sql")
plt.figure(figsize=(10, 5))
# Comparamos el % de volumen vs el % de ingresos para taxis amarillos
df_i6_yellow = df_i6[df_i6['tipo'] == 'yellow'].melt(
    id_vars=['mes'], 
    value_vars=['pct_viajes_aeropuerto', 'pct_ingresos_aeropuerto'],
    var_name='metrica', value_name='porcentaje'
)
sns.lineplot(data=df_i6_yellow, x="mes", y="porcentaje", hue="metrica", marker="^")
plt.title("Peso de los Aeropuertos: % Volumen vs % Ingresos (Yellow Taxis)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
